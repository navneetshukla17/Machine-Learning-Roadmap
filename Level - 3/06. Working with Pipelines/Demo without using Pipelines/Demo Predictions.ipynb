```python
import pickle
import numpy as np
```

```python
age_imputer = pickle.load(open('models/age_imputer.pkl', 'rb'))
clf = pickle.load(open('models/dt_clf.pkl', 'rb'))
embarked_imputer = pickle.load(open('models/embarked_imputer.pkl', 'rb'))
embarked_encoder = pickle.load(open('models/ohe_embarked.pkl', 'rb'))
sex_encoder = pickle.load(open('models/ohe_sex.pkl', 'rb'))
```

```python
test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'], dtype=object).reshape(1,7)
```

```python
test_input_embarked = embarked_encoder.transform(test_input[:,-1].reshape(1, 1))
test_input_sex = sex_encoder.transform(test_input[:,1].reshape(1, 1))
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_trf = np.concatenate((
    test_input[:,[0, 3, 4, 5]],
    test_input_age,
    test_input_sex,
    test_input_embarked
), axis=1)
```
