```python
import pandas as pd
import requests

url_template = "https://api.themoviedb.org/3/movie/top_rated?language=en-US&page={}"
headers = {
    "accept": "application/json",
    "Authorization": "Bearer eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiI1MjIyMGVjMmY4OWJkNjRhYWJiN2IwYzJkY2NiODg5ZCIsIm5iZiI6MTY3NDc1MDQ2NS45NjYsInN1YiI6IjYzZDJhYTAxNjZhZTRkMDA5ZTllY2Y2MyIsInNjb3BlcyI6WyJhcGlfcmVhZCJdLCJ2ZXJzaW9uIjoxfQ.k2K5I2-dU-ThOKM7HV4Ed_HQf3xLmRf7sWf_Ba2136c"
}

response = requests.get(url_template.format(1), headers=headers)
data = response.json()
total_pages = data.get('total_pages', 1)
print(f'Total pages available: {total_pages}')
```

```python
all_data = []

for i in range(1, total_pages + 1):
    print(f'Fetching page: {i}/{total_pages}...')
    url = url_template.format(i)
    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        results = response.json().get('results', [])
        if results:
            temp_df = pd.DataFrame(results)[
                            ["id", "original_title", "release_date", "overview", "popularity", "vote_average", "vote_count"]
            ]
            all_data.append(temp_df)
        else:
            print(f'Skipping page {i} (Status code: {response.status_code})')

df = pd.concat(all_data, ignore_index=True)
print(f'Total movies fetched: {len(df)}')
```

```python
df.to_csv('tmdb_top_rated_movies.csv', index=False)
```
