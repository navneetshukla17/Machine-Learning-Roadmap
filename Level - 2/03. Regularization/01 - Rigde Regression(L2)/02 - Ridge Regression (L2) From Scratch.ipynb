```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge
```

```python
X, y = make_regression(
    n_samples=500,
    n_features=1,
    n_informative=1,
    n_targets=1,
    noise=20,
    random_state=13
)
```

```python
plt.scatter(X, y, alpha=0.7)
plt.show()
```

```python
lr_reg = LinearRegression()
lr_reg.fit(X, y)
print('For Linear Regression Model')
print('-'*30)
print(f'Coefficient: {lr_reg.coef_}\nIntercept: {lr_reg.intercept_}')
```

```python
rr = Ridge(alpha=50)
rr.fit(X, y)
print('For Ridge Regression Model - Alpha=50')
print('-'*37)
print(f'Coefficient: {rr.coef_}\nIntercept: {rr.intercept_}')
```

```python
rr1 = Ridge(alpha=100)
rr1.fit(X, y)
print('For Ridge Regression Model - Alpha=100')
print('-'*37)
print(f'Coefficient: {rr.coef_}\nIntercept: {rr.intercept_}')
```

```python
plt.plot(X, y, 'o', alpha=0.7)
plt.plot(X, lr_reg.predict(X), color='red', label='Alpha=0')
plt.plot(X, rr.predict(X), color='black', label='Alpha=50')
plt.plot(X, rr1.predict(X), color='orange', label='Alpha=100')
plt.legend()
plt.show()
```
