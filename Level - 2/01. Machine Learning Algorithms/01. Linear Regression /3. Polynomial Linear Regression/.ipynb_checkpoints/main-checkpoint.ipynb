```python
# required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
```

```python
# generating random data for Simple Polynomial Linear Regression
X = 6 * np.random.rand(200, 1) - 3
y = 0.8 * X**2 + 0.9 * X + 2 + np.random.randn(200, 1) # formula: y = 0.8x^2 + 0.9x + 2
```

```python
plt.plot(X, y, '.')
plt.xlabel('X')
plt.ylabel('y')
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
LR = LinearRegression()
```

```python
LR.fit(X_train, y_train)
```

```python
y_pred = LR.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

```python
plt.plot(X_train, LR.predict(X_train), color='r')
plt.plot(X, y, 'b. ')
plt.xlabel('X')
plt.ylabel('y')
plt.show()
```

```python
poly = PolynomialFeatures(degree=2) # include bias parameter will remove the bias term.
```

```python
X_train_trans = poly.fit_transform(X_train)
X_test_trans = poly.transform(X_test)
```

```python
print(X_train[0])
print(X_train_trans[0])
```

```python
# now training with transformed data
LR = LinearRegression()
```

```python
LR.fit(X_train_trans, y_train)
```

```python
y_pred_trans = LR.predict(X_test_trans)
```

```python
r2_score(y_test, y_pred_trans)
```

```python
print(LR.coef_)
print(LR.intercept_)
```

```python
X_new = np.linspace(-3, 3, 200).reshape(200, 1)
X_new_poly = poly.transform(X_new)
y_pred_new = LR.predict(X_new_poly)
```

```python
plt.plot(X_new, y_pred_new, 'r-', linewidth=1, label='Predictions')
plt.plot(X_train, y_train, 'y.', label='Trainig points')
plt.plot(X_test, y_test, 'g.', label='Testing points')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
# polynomial regression function to test different degrees of the polynomial line
def polynomial_regression(degree):
    # generate random data with some noise
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)
    X_new_trans = poly.transform(X_new) 
    
    polynomial_features = PolynomialFeatures(degree=degree, include_bias=False)
    std_scaler = StandardScaler()
    lin_reg = LinearRegression()
    
    polynomial_pipe = Pipeline([
        ('polynomial_features', polynomial_features),
        ('scaler', std_scaler),
        ('linear_reg', lin_reg)
    ])
    
    polynomial_pipe.fit(X, y)
    y_pred_new = polynomial_pipe.predict(X_new)
    
    plt.plot(X_new, y_pred_new, 'r', label='Degree' + str(degree), linewidth=1)
    
    plt.plot(X_train, y_train, 'y.', linewidth=3)
    plt.plot(X_test, y_test, 'g.', linewidth=3)
    plt.axis([-3, 3, 0, 10])
    plt.legend()
    plt.show()
```

```python
polynomial_regression(degree=20)
```
