### Text Cleaning Process:
1. Sample `10,000` rows
2. Remove `HTML` tags
3. Remove `special` sharacteres
4. Converting every text to `lowercase`
5. Removing `stop` words (and, is, the, from etc)
6. Stemming 

### 1. Sample `10,000` rows 

```python
df = df.sample(10000)
df.shape
```

```python
df['sentiment'].replace({'positive':1, 'negative':0}, inplace=True)
df
```

### 2. Remove `HTML` tags

```python
import re
clean = re.compile('<.*?>')
re.sub(clean, '', df.iloc[2].review)
```

```python
def remove_html_tags(text):
    clean = re.compile('<.*?>')

    return re.sub(clean, '', text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
df['review']
```
