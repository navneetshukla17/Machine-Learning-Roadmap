```python
df = pd.read_csv('play_tennis.csv')
df.head()
```

```python
df.drop(columns=['day'], axis=1, inplace=True) # removing `day` col
df.head()
```

```python
# P(Yes) & P(No)
df.value_counts('play')
```

```python
# for temp col:
temp_cross_tab = pd.crosstab(df['temp'], df['play'])

P_of_cool_no = 1/5
P_of_hot_no = 2/5
P_of_mild_no = 2/5

P_of_cool_yes = 3/9
P_of_mild_yes = 4/9
P_of_hot_yes = 2/9

temp_cross_tab
```
