```python
class MyLR:
    """
    A simple linear regression class using OLS(ordinary least squares) technique.
    
    Methods:
    fit(Xtrain, ytrain): finds the value of m and b
    predict(Xtest): predicts output using y = m x + b formula
    """
    def __init__(self):
        """
        class initializer / constructor.
        It has arguments `Slope of the line (m)` and `y-Intercept (b)` which is initialy set to None.
        """
        self.m = None
        self.b = None

        
    def fit(self, X_train, y_train):
        """
        fits linear regression model to the given data.
        
        Parameters:
        Xtrain (int): training data (inputs)
        ytrain (int): training data (outputs)
        
        Returns:
        In this function we are just finding the value of `m` and `b`, It do not returns anything.
        """
        nume = 0
        deno = 0
        
        for i in range(X_train.shape[0]):
            nume = nume + ((X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean()))
            deno = deno + ((X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean()))
        
        self.m = nume / deno
        self.b = y_train.mean() - (self.m * X_train.mean())
        print(self.m)
        print(self.b)
        
    def predict(self, X_test):
        """
        predicts output for any given input.
        
        Parameters:
        Xtest (int): testing data
        
        Returns (int): prediction baised on the data it is fitted on
        """
        return self.m * X_test + self.b
```
