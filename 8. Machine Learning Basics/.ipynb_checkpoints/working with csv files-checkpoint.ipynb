### 13. Handling Dates

```python
pd.read_csv('IPL Matches 2008-2020.csv', parse_dates=['date']).info()
# converted date column dtype into actual date format 
```

### 14. Converters

```python
def rename(name):
    if name == 'Royal Challengers Bangalore':
        return 'RCB'
    elif name == 'Kings XI Punjab':
        return 'KX1P'

pd.read_csv('IPL Matches 2008-2020.csv', converters={'team1':rename})
```

### 15. na_values

```python
pd.read_csv('aug_train.csv', na_values='No relevent experience') # make any valeus as NAN
```

### 16. Loading huge dataset in small chunks

```python
dfs = pd.read_csv('aug_train.csv', chunksize=5000)

for chunks in dfs:
    print(chunks.shape)
```
